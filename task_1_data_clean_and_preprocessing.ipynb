```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python

from google.colab import files

# Upload the Kaggle dataset
uploaded = files.upload()

# Get uploaded filename
file_name = list(uploaded.keys())[0]

# Load dataset
df = pd.read_csv(file_name, sep='\t')

print("Dataset loaded successfully!")
```

```python

# Display first 5 rows
display(df.head())

# Display number of rows and columns
print("Dataset Shape:", df.shape)

# Display column names
print("\nColumn Names:")
print(df.columns.tolist())
```

```python

# 1. Check dataset information
print("DATASET INFORMATION")
df.info()

# 2. Statistical summary
print("\nSTATISTICAL SUMMARY")
display(df.describe(include='all'))

# 3. Check total rows and columns
print("\nDATASET DIMENSIONS")
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

# 4. Check column names
print("\nCOLUMN NAMES")
print(df.columns.tolist())
```

```python

# Count missing values in each column
missing_values = df.isnull().sum()

# Calculate missing value percentage
missing_percentage = (df.isnull().sum() / len(df)) * 100

# Create missing value report
missing_report = pd.DataFrame({
    'Missing Values': missing_values,
    'Missing Percentage': missing_percentage
})

display(missing_report)
```

```python

# Check missing values in Income
print("Missing values in Income:",
      df['Income'].isnull().sum())

# Check total missing values
print("\nTotal missing values:",
      df.isnull().sum().sum())

# Check duplicate records
print("\nDuplicate records:",
      df.duplicated().sum())
```

```python

# Calculate median income
median_income = df['Income'].median()

print("Median Income:", median_income)

# Fill missing Income values
df['Income'] = df['Income'].fillna(median_income)

# Verify missing values
print("\nMissing Income after cleaning:",
      df['Income'].isnull().sum())
```

```python

# Count duplicates before removal
duplicates_before = df.duplicated().sum()

print("Duplicates before cleaning:",
      duplicates_before)

# Remove duplicate records
df = df.drop_duplicates()

# Verify duplicates
duplicates_after = df.duplicated().sum()

print("Duplicates after cleaning:",
      duplicates_after)
```

```python

# Identify text columns
text_columns = df.select_dtypes(include=['object']).columns

# Remove extra spaces from text values
for col in text_columns:
    df[col] = df[col].str.strip()

print("Text spaces standardized successfully!")

# Display unique values from important columns
print("\nEducation values:")
print(df['Education'].unique())

print("\nMarital Status values:")
print(df['Marital_Status'].unique())
```

```python

# Convert date column
df['Dt_Customer'] = pd.to_datetime(
    df['Dt_Customer'],
    dayfirst=True,
    errors='coerce'
)

print("Date conversion completed!")

# Check data type
print("Date data type:", df['Dt_Customer'].dtype)

# Display sample dates
display(df[['Dt_Customer']].head())
```

```python

# Standardize column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r'[^a-zA-Z0-9]+', '_', regex=True)
    .str.strip('_')
)

print("Column names standardized successfully!")

print("\nUpdated Column Names:")
print(df.columns.tolist())
```

```python

print("FINAL DATA TYPES")

df.info()
```

```python

print("=" * 50)
print("FINAL DATA QUALITY REPORT")
print("=" * 50)

# Dataset dimensions
print("\n1. Dataset Dimensions")
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

# Missing values
print("\n2. Missing Values")
print(df.isnull().sum())

# Duplicate records
print("\n3. Duplicate Records")
print("Total Duplicates:", df.duplicated().sum())

# Data types
print("\n4. Data Types")
print(df.dtypes)

# Preview cleaned dataset
print("\n5. Cleaned Dataset Preview")
display(df.head())

print("\nData quality validation completed!")
```

```python

# Export cleaned dataset
df.to_csv(
    'cleaned_customer_personality.csv',
    index=False
)

print("Cleaned dataset saved successfully!")

# Download CSV file
from google.colab import files
files.download('cleaned_customer_personality.csv')
```
